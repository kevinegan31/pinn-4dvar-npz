```python
import pandas as pd

# Read the CSV
df = pd.read_csv("/Users/egank31/Documents/Documents - Otto/GitHub/pinns/npz/current/important_files_paper/data_analysis/current_files/hpo_training_times.csv")

# first 50 rows
df_50 = df.iloc[:50]

print(df_50.head())

```

```python
len(df_50)
```

```python
total_hours = df_50["runtime_hours"].sum()
total_minutes   = total_hours * 60
total_seconds = total_hours * 60 * 60

avg_hours     = total_hours / len(df)
avg_minutes   = total_minutes / len(df)

print(f"Total HPO wall clock time: {total_seconds:.2f} sec ({total_minutes:.2f} min, {total_hours:.2f} hrs)")
print(f"Average per trial: {avg_minutes:.2f} min ({avg_hours:.2f} hrs)")

print(f"Total HPO GPU time: {total_seconds*2:.2f} sec ({total_minutes*2:.2f} min, {total_hours*2:.2f} hrs)")
print(f"Average per trial: {avg_minutes*2:.2f} min ({avg_hours*2:.2f} hrs)")
```
